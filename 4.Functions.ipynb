```python
def Elegible():
    if gender == "male" and age >= 21:
        result = "ELIGIBLE"
    elif gender == "female" and age >= 18:
        result = "ELIGIBLE"
    else:
        result = "NOT ELIGIBLE"
    return result

gender = input("Enter your gender (male/female): ")
age = int(input("Enter your age: "))
print("Your Gender :", gender)
print("Your Age :", age)
eligibility = Elegible()
print("You are", eligibility)
```

```python
def  percentage(total,grand_total):
    return (total / grand_total) * 100
subject1 = int(input("Enter marks for subject 1: "))
subject2 = int(input("Enter marks for subject 2: "))
subject3 = int(input("Enter marks for subject 3: "))
subject4 = int(input("Enter marks for subject 4: "))
subject5 = int(input("Enter marks for subject 5: "))
grand_total = 500
total = subject1 + subject2 + subject3 + subject4 + subject5
percentage =  percentage(total, grand_total)
print("Subject 1:", subject1)
print("Subject 2:", subject2)
print("Subject 3:", subject3)
print("Subject 4:", subject4)
print("Subject 5:", subject5)
print("Total Marks:", total)
print("Percentage:", percentage)
```
