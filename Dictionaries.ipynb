dictionary are key value pair. key cannot be mutable and it's unique. values can be changeable(mutable)

```python
data = { "name": "Vinitha", "age": 30 }
print(data)
```

Creation of the dictionary.

```python
a = dict(a= 1, b= 2,c=3,d=4,e=5)
print(a)
```

Accessing of the data.  using [] ad get functionality

```python
print(data["name"])
print(data.get("age"))
```

addition and updation of the value

```python
data["Profession"] = "Engineer"
data["age"] = 25
print(data)
```

Deletion how many methods using 1.del keyword, 2. pop(keyname) 3. popitem(), 4. clear() to remove all the items

```python
del(a["a"])
print(a)
a.pop("c")
print(a)
a.popitem()
print(a)
a.clear()
print(a)
```

iterations how to use the dictionary in loops.

```python
for key in data:
    print(key)
for value in data.values():
    print(value)
for key, value in data.items():
    print(key, value)
```

nested dictionary

```python
data["work"] = {"1": "Engineer","2": "Developer"}
print(data)
print(data["work"]["1"])
```
